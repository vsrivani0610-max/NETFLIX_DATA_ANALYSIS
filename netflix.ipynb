### Handling Missing Values with Mean Imputation

```python
# Check for missing values in numerical columns that can be imputed by mean
numerical_cols = df.select_dtypes(include=np.number).columns
print("Missing values in numerical columns before imputation:")
print(df[numerical_cols].isnull().sum())

for col in numerical_cols:
    if df[col].isnull().any():
        mean_value = df[col].mean()
        df[col].fillna(mean_value, inplace=True)
        print(f"Filled missing values in '{col}' with its mean: {mean_value:.2f}")

print("\nMissing values in numerical columns after imputation:")
print(df[numerical_cols].isnull().sum())
```

### NumPy Array Operations

```python
# Convert the 'release_year' column to a numpy array
release_year_array = df['release_year'].to_numpy()

print("Numpy array from 'release_year' column (first 10 elements):")
print(release_year_array[:10])
print(f"\nType of the array: {type(release_year_array)}")
print(f"Shape of the array: {release_year_array.shape}")
```

```python
# Perform basic numpy operations
mean_year = np.mean(release_year_array)
std_dev_year = np.std(release_year_array)
min_year = np.min(release_year_array)
max_year = np.max(release_year_array)

print(f"Mean Release Year: {mean_year:.2f}")
print(f"Standard Deviation of Release Year: {std_dev_year:.2f}")
print(f"Minimum Release Year: {min_year}")
print(f"Maximum Release Year: {max_year}")
```

```python
print('--- NumPy Array Properties ---')

# Get the number of dimensions (ndim)
num_dimensions = release_year_array.ndim
print(f"Number of dimensions (ndim): {num_dimensions}")

# Get the shape of the array
array_shape = release_year_array.shape
print(f"Shape of the array: {array_shape}")

# Get the total number of elements (size)
array_size = release_year_array.size
print(f"Size of the array (total elements): {array_size}")
```

### Statistical Operations on Movie Duration using NumPy

```python
# Filter for movies and extract duration in minutes
movies_df = df[df['type'] == 'Movie'].copy()
movies_df['duration_minutes'] = movies_df['duration'].str.extract('(\d+)').astype(float)

# Convert to numpy array
movie_duration_array = movies_df['duration_minutes'].dropna().to_numpy()

print("Numpy array of movie durations (first 10 elements):")
print(movie_duration_array[:10])
print(f"\nType of the array: {type(movie_duration_array)}")
print(f"Shape of the array: {movie_duration_array.shape}")
```

```python
# Perform statistical operations on movie durations
mean_duration = np.mean(movie_duration_array)
median_duration = np.median(movie_duration_array)
std_dev_duration = np.std(movie_duration_array)
min_duration = np.min(movie_duration_array)
max_duration = np.max(movie_duration_array)

print(f"\nMean Movie Duration: {mean_duration:.2f} minutes")
print(f"Median Movie Duration: {median_duration:.2f} minutes")
print(f"Standard Deviation of Movie Duration: {std_dev_duration:.2f} minutes")
print(f"Minimum Movie Duration: {min_duration} minutes")
print(f"Maximum Movie Duration: {max_duration} minutes")
```

### Categorical Operations: Unique Values and Counts

```python
print('--- Unique Values and Counts for Categorical Columns ---')
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    print(f'\nColumn: {col}')
    if df[col].nunique() < 50: # Display unique values and counts if not too many
        print(df[col].value_counts())
    else:
        print(f'Number of unique values: {df[col].nunique()}')
        print(f'Top 10 unique values:\n{df[col].value_counts().head(10)}')
```

### NumPy Indexing and Slicing Operations

```python
print("Original release_year_array (first 15 elements):")
print(release_year_array[:15])

# 1. Basic Indexing: Accessing a single element
first_element = release_year_array[0]
print(f"\nFirst element: {first_element}")

fifth_element = release_year_array[4]
print(f"Fifth element: {fifth_element}")

# 2. Slicing: Accessing a range of elements
# Elements from index 2 up to (but not including) index 7
sliced_array = release_year_array[2:7]
print(f"\nElements from index 2 to 6: {sliced_array}")

# Elements from the beginning up to (but not including) index 5
sliced_from_start = release_year_array[:5]
print(f"Elements from start to index 4: {sliced_from_start}")

# Elements from index 10 to the end
sliced_to_end = release_year_array[10:]
print(f"Elements from index 10 to end (first 5 shown): {sliced_to_end[:5]}")
```

```python
# 3. Slicing with a step
# Every second element from index 0 to 10
stepped_slice = release_year_array[0:10:2]
print(f"\nEvery second element from index 0 to 9: {stepped_slice}")

# 4. Negative Indexing: Accessing elements from the end
last_element = release_year_array[-1]
print(f"\nLast element: {last_element}")

last_five_elements = release_year_array[-5:]
print(f"Last five elements: {last_five_elements}")

# 5. Conditional Indexing: Selecting elements based on a condition
# Find all release years after 2018
years_after_2018 = release_year_array[release_year_array > 2018]
print(f"\nRelease years after 2018 (first 10 shown): {years_after_2018[:10]}...")
print(f"Count of releases after 2018: {len(years_after_2018)}")
```

### Univariate Analysis of Columns

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot') # Use a consistent style for plots

# --- 1. Analysis of 'release_year' (Numerical) ---

# Histogram for release_year
plt.figure(figsize=(10, 6))
sns.histplot(df['release_year'], bins=30, kde=True)
plt.title('Distribution of Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.show()

# Box Plot for release_year
plt.figure(figsize=(10, 6))
sns.boxplot(x=df['release_year'])
plt.title('Box Plot of Release Year')
plt.xlabel('Release Year')
plt.show()
```

```python
# --- 2. Analysis of 'type' (Categorical) ---

plt.figure(figsize=(8, 5))
sns.countplot(x=df['type'], palette='viridis')
plt.title('Distribution of Content Type (Movie vs. TV Show)')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')
plt.show()
```

```python
# --- 3. Analysis of 'rating' (Categorical) ---

plt.figure(figsize=(12, 6))
sns.countplot(x=df['rating'], order=df['rating'].value_counts().index, palette='magma')
plt.title('Distribution of Content Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
```

```python
# --- 4. Analysis of 'date_added' (Date-Time) ---

# Convert 'date_added' to datetime objects
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

# Extract year and plot yearly counts
df['year_added'] = df['date_added'].dt.year.dropna().astype(int)

plt.figure(figsize=(12, 6))
sns.countplot(x=df['year_added'], order=df['year_added'].value_counts().sort_index().index, palette='plasma')
plt.title('Number of Titles Added Over Years')
plt.xlabel('Year Added')
plt.ylabel('Number of Titles')
plt.xticks(rotation=70)
plt.show()
```

```python
# --- 5. Analysis of 'country' (Multi-value Categorical) ---

# Split and explode countries, then get top 10
top_countries = (
    df['country']
    .dropna()
    .str.split(', ')
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(12, 7))
sns.barplot(x=top_countries.values, y=top_countries.index, palette='cividis')
plt.title('Top 10 Countries Producing Content')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.show()
```

```python
# --- 6. Analysis of 'listed_in' (Genres, Multi-value Categorical) ---

# Split and explode genres, then get top 10
top_genres = (
    df['listed_in']
    .dropna()
    .str.split(', ')
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(12, 7))
sns.barplot(x=top_genres.values, y=top_genres.index, palette='inferno')
plt.title('Top 10 Genres on Netflix')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
plt.show()
```

### Multivariate Analysis of Columns

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')

# Multivariate Analysis

# 1. Content Type vs. Release Year
plt.figure(figsize=(12, 7))
sns.boxplot(x='type', y='release_year', data=df, palette='coolwarm')
plt.title('Release Year Distribution by Content Type')
plt.xlabel('Content Type')
plt.ylabel('Release Year')
plt.show()

# 2. Content Type vs. Rating
plt.figure(figsize=(14, 7))
sns.countplot(data=df, x='rating', hue='type', order=df['rating'].value_counts().index, palette='viridis')
plt.title('Content Ratings by Type')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.legend(title='Content Type')
plt.show()

# 3. Content Type vs. Year Added
# Ensure 'year_added' is available from previous steps
if 'year_added' in df.columns:
    plt.figure(figsize=(15, 7))
    sns.countplot(data=df, x='year_added', hue='type', palette='magma', order=df['year_added'].value_counts().sort_index().index)
    plt.title('Number of Titles Added Each Year by Content Type')
    plt.xlabel('Year Added')
    plt.ylabel('Number of Titles')
    plt.xticks(rotation=70)
    plt.legend(title='Content Type')
    plt.show()
else:
    print("'year_added' column not found. Please ensure 'date_added' parsing and 'year_added' extraction were executed.")

# 4. Top Countries by Content Type
# Prepare data for countplot of top countries by type
country_temp_df = df[['type', 'country']].dropna(subset=['country']).copy()
country_temp_df['country_split'] = country_temp_df['country'].str.split(', ')
country_exploded_df = country_temp_df.explode('country_split')

# Get top 10 countries for ordering the plot
top_10_countries_plot = country_exploded_df['country_split'].value_counts().head(10).index

# Filter the exploded DataFrame to include only the top 10 countries
filtered_country_type_counts = country_exploded_df[country_exploded_df['country_split'].isin(top_10_countries_plot)]

plt.figure(figsize=(14, 8))
sns.countplot(data=filtered_country_type_counts, y='country_split', hue='type', order=top_10_countries_plot, palette='cividis')
plt.title('Top 10 Countries: Content Type Distribution')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.legend(title='Content Type')
plt.show()

# 5. Top Genres by Content Type
# Prepare data for countplot of top genres by type
genre_temp_df = df[['type', 'listed_in']].dropna(subset=['listed_in']).copy()
genre_temp_df['listed_in_split'] = genre_temp_df['listed_in'].str.split(', ')
genre_exploded_df = genre_temp_df.explode('listed_in_split')

# Get top 10 genres for ordering the plot
top_10_genres_plot = genre_exploded_df['listed_in_split'].value_counts().head(10).index

# Filter the exploded DataFrame to include only the top 10 genres
filtered_genre_type_counts = genre_exploded_df[genre_exploded_df['listed_in_split'].isin(top_10_genres_plot)]

plt.figure(figsize=(14, 8))
sns.countplot(data=filtered_genre_type_counts, y='listed_in_split', hue='type', order=top_10_genres_plot, palette='inferno')
plt.title('Top 10 Genres: Content Type Distribution')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
plt.legend(title='Content Type')
plt.show()
```
